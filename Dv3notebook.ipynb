Imports:

```python
from dice import Dice
from items import Weapon 
from character import Character, Warrior, Mage
from battle import Battle 
```

Presentation of Dice class:

```python
d6 = Dice()
d20 = Dice(sides=20)

print("Default dice sides:", d6.sides) 
print("Shows dice with decided sides:", repr(d20))

for i in range(1,4):
    result = d20.roll()
    print("With .roll(): You got", result, "on the dice")

for i in range(1,4): 
    result = d20()
    print("With __call__ : You got", result, "on the dice")
```

Presentation of Weapon class:

```python
sword = Weapon(name="sword", damage_dice=Dice(sides=14))
print("Weapon name:", sword.name)

for i in range(1,4):
    print("Weapon damage based on dice roll:", sword.roll_damage())
```

Presentation of Character class:

```python
hero = Character(name="Jim", hp_max=100, damage=8, weapon=None)
print(hero)
print("Is the character alive?",hero.is_alive)

damg = 25
hero.take_damage(damg)
print(f"The characters hp after taking {damg} damage: {hero.hp}")

hl = 40
hero.heal(hl)
print(f"The characters hp after healing {hl} hp:{hero.hp}") 

dmg = 250
hero.take_damage(dmg)
print(f"The characters hp after taking {dmg} damage is:{hero.hp} and is the character alive? {hero.is_alive}")
```

Simulated battles: 
1st battle, one with weapon, one without. Sword gives random damage, which is decided by the Dice.

```python
sword = Weapon(name="sword", damage_dice=Dice(sides=14))
character1 = Character(name="Micheal", hp_max=100, damage=10)
character2 = Character(name="Jim", hp_max = 100, damage=12, weapon=sword)

battle1 = Battle(hero=character1, enemy=character2)
battle1.simulate_battle() 
```

2nd battle: 
Warrior with sword vs normal character with dagger

```python
sword = Weapon(name="sword", damage_dice=Dice(sides=15))
dagger = Weapon(name="dagger", damage_dice=Dice(sides=20))

warrior = Warrior(name="Pam", hp_max=100, damage=10, weapon=sword)
character2 = Character(name="Jim", hp_max = 100, damage=12, weapon=dagger)

battle2 = Battle(hero=warrior, enemy=character2)
battle2.simulate_battle() 
```

3rd Battle, Warrior vs Mage 

```python
sword = Weapon(name="sword", damage_dice=Dice(sides=15))
staff = Weapon(name="staff", damage_dice=Dice(sides=15))

warrior = Warrior(name="Pam", hp_max=100, damage=10, weapon=sword)
mage = Mage(name="Angela", hp_max = 90, damage=10, mana=30, weapon=staff)

battle3 = Battle(hero=warrior, enemy=mage)
battle3.simulate_battle() 
```
