Presentation of Dice class:

```python
d6 = Dice()
d20 = Dice(sides=20)

print("Default dice sides:", d6.sides) 
print("Shows dice with decided sides:", repr(d20))

for i in range(1,4):
    result = d20.roll()
    print("With .roll(): You got", result, "on the dice")

for i in range(1,4): 
    result = d20()
    print("With __call__ : You got", result, "on the dice")
```

Presentation of Weapon class:

```python
sword = Weapon(name="sword", damage_dice=Dice(sides=14))
print("Weapon name:", sword.name)

for i in range(1,4):
    print("Weapon damage based on dice roll:", sword.roll_damage())
```

Presentation of Character class:

```python
hero = Character(name="Jim", hp_max=100, damage=8, weapon=None)
print(hero)
print("Is the character alive?",hero.is_alive)

damg = 25
hero.take_damage(damg)
print(f"The characters hp after taking {damg} damage: {hero.hp}")

hl = 40
hero.heal(hl)
print(f"The characters hp after healing {hl} hp:{hero.hp}") 

dmg = 250
hero.take_damage(dmg)
print(f"The characters hp after taking {dmg} damage is:{hero.hp} and is the character alive? {hero.is_alive}")
```
